```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing, load_breast_cancer

# --- PART 1: Data Collection ---
# California Housing Data load karna
housing = fetch_california_housing()
df_h = pd.DataFrame(housing.data, columns=housing.feature_names)
df_h['Price'] = housing.target

# Breast Cancer (Biological) Data load karna
cancer = load_breast_cancer()
df_b = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df_b['Target'] = cancer.target

# --- PART 2: Data Preprocessing ---
# Check for missing values and duplicates
print(f"Housing Nulls: {df_h.isnull().sum().sum()}")
df_h.drop_duplicates(inplace=True)

print(f"Biological Nulls: {df_b.isnull().sum().sum()}")
df_b.drop_duplicates(inplace=True)

# --- PART 3 & 4: Normalization (Log 2) & Visualization ---
plt.figure(figsize=(12, 5))

# Before Normalization
plt.subplot(1, 2, 1)
sns.histplot(df_h['MedInc'], kde=True, color='blue')
plt.title("Before Log 2 Normalization (MedInc)")

# Applying Log 2 Normalization
df_h['MedInc_Log'] = np.log2(df_h['MedInc'] + 1)

# After Normalization
plt.subplot(1, 2, 2)
sns.histplot(df_h['MedInc_Log'], kde=True, color='green')
plt.title("After Log 2 Normalization (MedInc)")
plt.show()

# --- PART 5: Exploratory Data Analysis (EDA) ---

# 1. Housing Dataset: Correlation Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df_h.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Housing Correlation Matrix")
plt.show()

# 2. Biological Dataset: Count Plot (Malignant vs Benign)
plt.figure(figsize=(6, 4))
sns.countplot(x='Target', data=df_b, palette='magma')
plt.xticks(ticks=[0, 1], labels=['Malignant', 'Benign'])
plt.title("Cancer Diagnosis Count")
plt.show()

# --- WRITE 5 INSIGHTS ---
print("\n" + "="*20 + " 5 KEY INSIGHTS " + "="*20)
print("1. Housing: Median Income (MedInc) has the strongest positive correlation with House Price.")
print("2. Housing: Log 2 normalization effectively reduced the skewness in the income distribution.")
print("3. Biological: The dataset contains more Benign (1) cases than Malignant (0) cases.")
print("4. Biological: Features like 'mean radius' are critical indicators for cancer classification.")
print("5. General: Preprocessing confirmed that both datasets were clean with no missing values.")
```
